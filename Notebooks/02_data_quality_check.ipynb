```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:qcommerce123@localhost:5432/saas_health')

customers_df = pd.read_sql("SELECT * FROM customers", engine)
subscriptions_df = pd.read_sql("SELECT * FROM subscriptions", engine)
usage_events_df = pd.read_sql("SELECT * FROM usage_events", engine)
payments_df = pd.read_sql("SELECT * FROM payments", engine)

print("Data reloaded from PostgreSQL")
print(f"Customers: {len(customers_df)}, Subscriptions: {len(subscriptions_df)}, "
      f"Usage events: {len(usage_events_df)}, Payments: {len(payments_df)}")
```
